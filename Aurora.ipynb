```python
df['timestamp'] = pd.to_datetime(df['timestamp'])
```

```python
def verificar_decolagem(row):
    # Condições de sucesso (True se for uma falha)
    falhas = []

    # Condição: temperatura_interna_c < 25 (sucesso) -> Falha se >= 25
    if row['temperatura_interna_c'] >= 25:
        falhas.append('Temperatura Interna Elevada')

    # Condição: temperatura_externa_c < -100 (sucesso) -> Falha se >= -100
    if row['temperatura_externa_c'] >= -100:
        falhas.append('Temperatura Externa Insuficiente')

    # Condição: integridade_estrutural != 0 (falso) -> Falha se == 0
    if row['integridade_estrutural'] == 0:
        falhas.append('Integridade Estrutural Comprometida')

    # Condição: nivel_energia_pct > 100 (falso) -> Falha se > 100
    if row['nivel_energia_pct'] > 100 or row['nivel_energia_pct'] <= 0: # Adicionei <=0 para ter um limite inferior mais realista
        falhas.append('Nível de Energia Inválido')

    # Condição: pressao_tanques_bar < 240 (falso) -> Falha se < 240
    if row['pressao_tanques_bar'] < 240:
        falhas.append('Pressão dos Tanques Baixa')

    # Condição: status_modulos_criticos != ok (falso) -> Falha se != 'OK'
    if row['status_modulos_criticos'] != 'OK':
        falhas.append(f"Status dos Módulos Críticos: {row['status_modulos_criticos']}")

    # Se não houver falhas, é sucesso
    if not falhas:
        return 'sucesso'
    else:
        return f"falhou: {', '.join(falhas)}"

df['decolagem_status'] = df.apply(verificar_decolagem, axis=1)
```

#calculo energico


```python
def calcular_autonomia(capacidade_total_kwh, carga_atual_pct, consumo_decolagem_kwh, perdas_pct):

    # Energia disponível
    energia_disponivel = capacidade_total_kwh * (carga_atual_pct / 100)

    # Aplicando perdas
    energia_util = energia_disponivel * (1 - perdas_pct / 100)

    # Autonomia após decolagem
    autonomia = energia_util - consumo_decolagem_kwh

    return autonomia


# Exemplo de uso
capacidade = 1000      # kWh
carga = 100           # %
consumo_decolagem = 200  # kWh
perdas = 10           # %

autonomia = calcular_autonomia(capacidade, carga, consumo_decolagem, perdas)

print(f"Autonomia restante: {autonomia:.2f} kWh")
```

#analise assistida por IA

```python
import csv

def classificar(linha):
    temp = float(linha["temperatura_interna_c"])
    integridade = int(linha["integridade_estrutural"])
    energia = float(linha["nivel_energia_pct"])
    pressao = float(linha["pressao_tanques_bar"])
    status = linha["status_modulos_criticos"]

    # CRÍTICO
    if integridade == 0 or energia < 70 or pressao < 220 or pressao > 280:
        return "CRITICO"

    # ALERTA
    if (70 <= energia < 90 or
        220 <= pressao < 240 or
        260 < pressao <= 280 or
        status != "OK"):
        return "ALERTA"

    # NORMAL
    return "NORMAL"


def detectar_anomalias(linha_atual, linha_anterior):
    anomalias = []

    if linha_anterior:
        energia_atual = float(linha_atual["nivel_energia_pct"])
        energia_anterior = float(linha_anterior["nivel_energia_pct"])

        # queda brusca de energia
        if energia_anterior - energia_atual > 5:
            anomalias.append("Queda brusca de energia")

        pressao_atual = float(linha_atual["pressao_tanques_bar"])
        pressao_anterior = float(linha_anterior["pressao_tanques_bar"])

        # queda contínua de pressão
        if pressao_atual < pressao_anterior:
            anomalias.append("Possível vazamento (pressão caindo)")

    return anomalias


def sugerir_riscos(linha):
    riscos = []

    energia = float(linha["nivel_energia_pct"])
    pressao = float(linha["pressao_tanques_bar"])
    integridade = int(linha["integridade_estrutural"])
    status = linha["status_modulos_criticos"]

    if energia < 70:
        riscos.append("Risco de falha energética")

    if integridade == 0:
        riscos.append("Risco estrutural crítico")

    if pressao < 220:
        riscos.append("Possível vazamento")

    if status != "OK":
        riscos.append("Falha em módulos críticos")

    return riscos


# Leitura do CSV
with open("telemetria.csv", newline="", encoding="utf-8") as arquivo:
    leitor = csv.DictReader(arquivo)

    anterior = None

    for linha in leitor:
        classe = classificar(linha)
        anomalias = detectar_anomalias(linha, anterior)
        riscos = sugerir_riscos(linha)

        print(f"\n--- {linha['timestamp']} ---")
        print(f"Classificação: {classe}")

        if anomalias:
            print("Anomalias:", ", ".join(anomalias))

        if riscos:
            print("Riscos:", ", ".join(riscos))

        anterior = linha
```
